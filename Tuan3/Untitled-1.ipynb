## 2. GIẢI THÍCH BÀI LÀM

### 2.1. Cơ sở lý thuyết
LINQ (Language-Integrated Query) là một tính năng mạnh mẽ của C#, cho phép truy vấn dữ liệu trực tiếp trong ngôn ngữ lập trình. LINQ hỗ trợ hai cú pháp chính:
- **Query Syntax:** Cú pháp giống SQL, sử dụng các từ khóa `from`, `where`, `select`, `group by`, `join`.
- **Method Syntax:** Sử dụng các phương thức mở rộng như `Where()`, `Select()`, `OrderBy()`, `GroupBy()`, `Join()`.

### 2.2. Giải thích chi tiết từng bài

**Bài 2.1 – Truy vấn mảng số nguyên:**
- Sử dụng `Where()` để lọc các phần tử chia hết cho cả 4 và 3 (tương đương chia hết cho 12).
- Sử dụng `Where()` để lọc các phần tử nhỏ hơn hoặc bằng 3.
- Sử dụng `Select()` kết hợp toán tử ba ngôi để biến đổi dãy: nếu phần tử là số chẵn thì chia đôi, nếu là số lẻ thì giữ nguyên.

**Bài 2.2 – Truy vấn mảng chuỗi:**
- Sử dụng `Where()` kết hợp `Length` để lọc các từ có đúng 4 ký tự, sau đó `OrderBy()` để sắp xếp theo ký tự đầu tiên.
- Sử dụng `Select()` để biến đổi mỗi phần tử thành dạng `<chữ thường> - <CHỮ HOA>`.
- Sử dụng `Where()` kết hợp `Contains()` để tìm các từ chứa ký tự "ư".
- Sử dụng `Where()` kết hợp `char.IsUpper()` để lọc các từ bắt đầu bằng chữ in hoa.

**Bài 3.1 – Thống kê mảng số nguyên:**
- Sử dụng `Length` để đếm tổng số phần tử, `Count()` để đếm số chẵn và số lẻ.
- Sử dụng `Sum()`, `Min()`, `Max()` để tính tổng, giá trị nhỏ nhất và lớn nhất.
- Sử dụng `Distinct().Count()` để đếm số lượng giá trị khác nhau.
- Sử dụng `GroupBy()` để phân nhóm các phần tử theo số dư khi chia cho 5.

**Bài 3.2 – Thống kê mảng chuỗi:**
- Sử dụng `Min()` và `Max()` trên thuộc tính `Length` để tìm độ dài ngắn nhất và dài nhất, sau đó lọc các phần tử tương ứng.
- Sử dụng `GroupBy()` kết hợp `Split(' ')[0]` để phân nhóm theo từ đầu tiên của mỗi món ăn.
- Sử dụng `Count()` kết hợp `StartsWith()` để đếm số món có từ đầu tiên là "Bánh".

**Bài 4.1 – Xây dựng lớp MonHoc:**
- Khai báo lớp `MonHoc` với 4 thuộc tính: `MaMon` (string), `TenMon` (string), `He` (string), `SoTiet` (byte).

- Sử dụng `Where()`, `OrderByDescending()` và `ThenBy()` để lọc và sắp xếp các môn thuộc hệ "CD" theo số tiết giảm dần, sau đó theo mã môn tăng dần.
- Sử dụng `Where()` kết hợp `Contains()` để lọc các môn có tên chứa "web".
- Sử dụng `Where()` và `OrderBy()` để lọc các môn thuộc hệ "KTV" và sắp xếp theo mã môn.

**Bài 5.2 – Thống kê trên List<MonHoc>:**
- Sử dụng `Count()` để đếm tổng số môn và số môn bắt đầu bằng "Lập trình".
- Sử dụng `Where()` kết hợp `Sum()` để tính tổng số tiết của hệ "KTV".
- Sử dụng `GroupBy()` để nhóm các môn theo hệ, theo số tiết.

- Sử dụng `GroupBy()` kết hợp `Select()` để tính toán thống kê chi tiết theo từng hệ (tổng số môn, tổng số tiết, max, min).

- Sử dụng `FirstOrDefault()` để tìm môn học đầu tiên bắt đầu bằng "Lập trình".
- Sử dụng `GroupBy()` kết hợp vòng lặp để liệt kê các môn theo từng hệ và đánh số thứ tự.
