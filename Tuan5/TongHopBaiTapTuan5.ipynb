    "using System;\n",
    "using System.Windows.Forms;\n",
    "\n",
    "namespace Bai3\n",
    "{\n",
    "    public partial class Form1 : Form\n",
    "    {\n",
    "        public Form1()\n",
    "        {\n",
    "            InitializeComponent();\n",
    "        }\n",
    "\n",
    "        private void Form1_Load(object sender, EventArgs e)\n",
    "        {\n",
    "            for (char c = 'A'; c <= 'Z'; c++)\n",
    "                treeView1.Nodes.Add(c.ToString());\n",
    "        }\n",
    "\n",
    "        private void btnAdd_Click(object sender, EventArgs e)\n",
    "        {\n",
    "            string firstName = txtFirstName.Text.Trim();\n",
    "            string lastName = txtLastName.Text.Trim();\n",
    "\n",
    "            if (string.IsNullOrEmpty(firstName) || string.IsNullOrEmpty(lastName))\n",
    "            {\n",
    "                MessageBox.Show(\"Vui lòng nhập đầy đủ First Name và Last Name!\");\n",
    "                return;\n",
    "            }\n",
    "\n",
    "            string firstLetter = firstName.Substring(0, 1).ToUpper();\n",
    "            TreeNode parentNode = null;\n",
    "\n",
    "            foreach (TreeNode node in treeView1.Nodes)\n",
    "            {\n",
    "                if (node.Text == firstLetter)\n",
    "                {\n",
    "                    parentNode = node;\n",
    "                    break;\n",
    "                }\n",
    "            }\n",
    "\n",
    "            if (parentNode != null)\n",
    "            {\n",
    "                parentNode.Nodes.Add($\"{firstName}, {lastName}\");\n",
    "                parentNode.Expand();\n",
    "                txtFirstName.Clear();\n",
    "                txtLastName.Clear();\n",
    "                txtFirstName.Focus();\n",
    "            }\n",
    "            else\n",
    "            {\n",
    "                MessageBox.Show(\"Chữ cái đầu không hợp lệ (phải A-Z)!\");\n",
    "            }\n",
    "        }\n",
    "\n",
    "        private void btnExit_Click(object sender, EventArgs e)\n",
    "        {\n",
    "            if (MessageBox.Show(\"Bạn có chắc chắn muốn thoát?\", \"Xác nhận\", MessageBoxButtons.YesNo) == DialogResult.Yes)\n",
    "                Application.Exit();\n",
