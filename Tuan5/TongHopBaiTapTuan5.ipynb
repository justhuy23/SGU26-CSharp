```python
using System;
using System.Windows.Forms;

using System;
using System.Windows.Forms;

namespace Bai1
{
    public partial class Form1 : Form
    {
        public Form1()
        {
            InitializeComponent();
        }

        private bool IsPrime(int n)
        {
            if (n < 2) return false;
            for (int i = 2; i <= Math.Sqrt(n); i++)
            {
                if (n % i == 0) return false;
            }
            return true;
        }

        private void btnCapNhat_Click(object sender, EventArgs e)
        {
            if (!int.TryParse(txtNhapSo.Text, out int number) || number <= 0)
            {
                MessageBox.Show("Vui lòng nhập một số nguyên dương hợp lệ!", "Lỗi",
                    MessageBoxButtons.OK, MessageBoxIcon.Error);
                txtNhapSo.Clear();
                txtNhapSo.Focus();
                return;
            }

            if (cboSo.Items.Contains(number))
            {
                MessageBox.Show("Số này đã tồn tại trong danh sách!", "Thông báo",
                    MessageBoxButtons.OK, MessageBoxIcon.Warning);
            }
            else
            {
                cboSo.Items.Add(number);
            }

            txtNhapSo.Clear();
            txtNhapSo.Focus();
        }

        private void cboSo_SelectedIndexChanged(object sender, EventArgs e)
        {
            if (cboSo.SelectedItem == null) return;

            int selectedNumber = (int)cboSo.SelectedItem;
            lstUocSo.Items.Clear();

            for (int i = 1; i <= selectedNumber; i++)
            {
                if (selectedNumber % i == 0)
                    lstUocSo.Items.Add(i);
            }
        }

        private void btnTong_Click(object sender, EventArgs e)
        {
            int sum = 0;
            foreach (int item in lstUocSo.Items)
                sum += item;

            MessageBox.Show($"Tổng các ước số là: {sum}", "Kết quả");
        }

        private void btnChan_Click(object sender, EventArgs e)
        {
            int count = 0;
            foreach (int item in lstUocSo.Items)
            {
                if (item % 2 == 0) count++;
            }

            MessageBox.Show($"Số lượng các ước số chẵn là: {count}", "Kết quả");
        }

        private void btnNguyenTo_Click(object sender, EventArgs e)
        {
            int count = 0;
            foreach (int item in lstUocSo.Items)
            {
                if (IsPrime(item)) count++;
            }

            MessageBox.Show($"Số lượng các ước số nguyên tố là: {count}", "Kết quả");
        }

        private void btnThoat_Click(object sender, EventArgs e)
        {
            if (MessageBox.Show("Bạn có chắc chắn muốn thoát?", "Xác nhận",
                MessageBoxButtons.YesNo, MessageBoxIcon.Question) == DialogResult.Yes)
            {
                Application.Exit();
            }
        }
    }
}
```
