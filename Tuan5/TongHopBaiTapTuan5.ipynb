## BÀI 1: CHƯƠNG TRÌNH TÍNH TOÁN ƯỚC SỐ

| Tên hàm | Loại | Chức năng |
|---------|------|-----------|
| `Form1()` | Constructor | Khởi tạo Form, gọi `InitializeComponent()` để vẽ giao diện. |
| `IsPrime(int n)` | Hàm phụ trợ (private) | Kiểm tra số `n` có phải số nguyên tố không. Trả về `true` nếu là số nguyên tố, ngược lại `false`. Duyệt từ 2 đến `√n`, nếu `n` chia hết cho bất kỳ số nào thì không phải nguyên tố. |
| `btnCapNhat_Click` | Sự kiện Click | Xử lý khi nhấn nút **Cập nhật**. Kiểm tra dữ liệu nhập (phải là số nguyên dương), kiểm tra trùng lặp trong ComboBox, thêm số mới vào ComboBox, xóa TextBox và đặt lại con trỏ. |
| `cboSo_SelectedIndexChanged` | Sự kiện | Kích hoạt khi người dùng chọn một số trong ComboBox. Xóa ListBox cũ, sau đó duyệt từ 1 đến số được chọn để tìm và hiển thị tất cả ước số vào ListBox. |
| `btnTong_Click` | Sự kiện Click | Tính tổng tất cả các ước số đang có trong ListBox và hiển thị kết quả bằng MessageBox. |
| `btnChan_Click` | Sự kiện Click | Đếm số lượng ước số là số chẵn (chia hết cho 2) trong ListBox và hiển thị kết quả. |
| `btnNguyenTo_Click` | Sự kiện Click | Đếm số lượng ước số là số nguyên tố (gọi hàm `IsPrime`) trong ListBox và hiển thị kết quả. |
| `btnThoat_Click` | Sự kiện Click | Hiển thị hộp thoại xác nhận Yes/No. Nếu chọn Yes thì gọi `Application.Exit()` để thoát chương trình. |

## BÀI 2: CHƯƠNG TRÌNH XỬ LÝ LISTBOX

| Tên hàm | Loại | Chức năng |
|---------|------|-----------|
| `Form1()` | Constructor | Khởi tạo Form, gọi `InitializeComponent()`. |
| `btnNhap_Click` | Sự kiện Click | Kiểm tra dữ liệu nhập từ TextBox có phải số nguyên không. Nếu hợp lệ thì thêm vào ListBox, sau đó xóa TextBox và đặt lại con trỏ. |
| `btnTong_Click` | Sự kiện Click | Duyệt toàn bộ phần tử trong ListBox, chuyển sang kiểu `int` và cộng dồn để tính tổng. Hiển thị kết quả bằng MessageBox. |
| `btnXoaDauCuoi_Click` | Sự kiện Click | Xóa phần tử đầu tiên (index 0). Nếu ListBox vẫn còn phần tử thì tiếp tục xóa phần tử cuối cùng (index `Count - 1`). |
| `btnXoaChon_Click` | Sự kiện Click | Kiểm tra xem có phần tử nào đang được chọn không. Nếu có, duyệt ngược từ cuối danh sách `SelectedIndices` để xóa từng phần tử được chọn (duyệt ngược để tránh lỗi index thay đổi). |
| `btnTang2_Click` | Sự kiện Click | Duyệt từng phần tử trong ListBox, cộng thêm 2 đơn vị và gán lại giá trị mới vào vị trí đó. |
| `btnBinhPhuong_Click` | Sự kiện Click | Duyệt từng phần tử trong ListBox, tính bình phương (`val * val`) và gán lại giá trị mới. |
| `btnChan_Click` | Sự kiện Click | Xóa tất cả lựa chọn cũ (`ClearSelected`), sau đó duyệt toàn bộ ListBox và chọn (`SetSelected`) những phần tử là số chẵn. |
| `btnLe_Click` | Sự kiện Click | Tương tự `btnChan_Click` nhưng chọn những phần tử là số lẻ. |
| `btnKetThuc_Click` | Sự kiện Click | Hiển thị hộp thoại xác nhận. Nếu đồng ý thì thoát chương trình. |

## BÀI 3: CHƯƠNG TRÌNH QUẢN LÝ DANH BẠ (TREEVIEW)

| Tên hàm | Loại | Chức năng |
|---------|------|-----------|
| `Form1()` | Constructor | Khởi tạo Form, gọi `InitializeComponent()`. |
| `Form1_Load` | Sự kiện Load | Kích hoạt khi Form bắt đầu hiển thị. Dùng vòng lặp từ 'A' đến 'Z' để tạo 26 node cha trong TreeView (mỗi node là một chữ cái). |
| `btnAdd_Click` | Sự kiện Click | Lấy First Name và Last Name từ TextBox. Kiểm tra rỗng. Lấy chữ cái đầu của First Name, chuyển thành chữ hoa. Duyệt các node trong TreeView để tìm node cha tương ứng. Nếu tìm thấy, thêm tên đầy đủ (`FirstName, LastName`) vào node con, mở rộng node cha và xóa trắng TextBox. Nếu không tìm thấy (ký tự không phải A-Z), hiển thị thông báo lỗi. |
| `btnExit_Click` | Sự kiện Click | Hiển thị hộp thoại xác nhận Yes/No. Nếu chọn Yes thì gọi `Application.Exit()` để thoát. |

## TÓM TẮT KỸ THUẬT CHUNG

| Kỹ thuật | Mô tả | Bài áp dụng |
|----------|-------|-------------|
| **Kiểm tra dữ liệu** | Dùng `int.TryParse` để kiểm tra nhập liệu có phải số nguyên | Bài 1, 2 |
| **Xử lý sự kiện** | Gán sự kiện `Click`, `SelectedIndexChanged`, `KeyDown`, `Load` cho control | Cả 3 bài |
| **MessageBox** | Hiển thị thông báo, kết quả, xác nhận | Cả 3 bài |
| **Duyệt tập hợp** | Dùng `foreach` hoặc vòng lặp `for` để duyệt ListBox, ComboBox, TreeView | Cả 3 bài |
| **Ép kiểu** | `int.Parse()`, `(int)`, `.ToString()` để chuyển đổi dữ liệu | Bài 1, 2 |
| **Xử lý nhiều lựa chọn** | `SelectionMode = MultiExtended` và `SelectedIndices` | Bài 2 |
| **Duyệt ngược khi xóa** | Duyệt từ `Count - 1` về 0 để tránh lỗi index | Bài 2 |
| **Tìm kiếm node** | Duyệt `TreeView.Nodes` để tìm node cha theo chữ cái | Bài 3 |
| **Thêm node động** | `parentNode.Nodes.Add()` để thêm node con | Bài 3 |
| **Mở rộng node** | `parentNode.Expand()` để hiển thị node con | Bài 3 |
